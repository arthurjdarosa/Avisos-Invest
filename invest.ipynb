```python
# Imports
import os
from fastapi import FastAPI
import yfinance as yf
import requests
from dotenv import load_dotenv
```

```python
TOKEN = os.getenv("TELEGRAM_TOKEN")
CHAT_ID = os.getenv("TELEGRAM_CHAT_ID")
```

```python
load_dotenv()
def enviar_mensagem_telegram(mensagem):
    url = f"https://api.telegram.org/bot{TOKEN}/sendMessage"
    
    payload = {
        "chat_id": CHAT_ID,
        "text": mensagem
    }
    
    response = requests.post(url, json=payload)
    return response.json()

# --- EXECUÇÃO MANUAL ---
if __name__ == "__main__":
    print("Tentando enviar mensagem...")
    resultado = enviar_mensagem_telegram("Teste")
    print(f"Resposta do Telegram: {resultado}")
```

```python
# AMBIENTE
app = FastAPI()
```

```python
TOKEN = os.getenv("TELEGRAM_TOKEN")
CHAT_ID = os.getenv("TELEGRAM_CHAT_ID")

def enviar_mensagem_telegram(mensagem):
    url = f"https://api.telegram.org/bot{TOKEN}/sendMessage?chat_id={CHAT_ID}&text={mensagem}"
    requests.get(url)
```

```python
@app.get("/check-invest")
def check_investiments():
    ticker = "MXRF11.SA"
    dados = yf.download(ticker, period="6mo")
    
    # CALCULOS
    preco_atual = dados['Close'].iloc[-1]
    media_hoje = dados['Close'].rolling(window=90).mean ().iloc[-1]

    if preco_atual > media_hoje:
        tlg_mensagem = f"Preço R${preco_atual:.2f}\n Media: R${media_hoje:.2f}"
        enviar_mensagem_telegram(tlg_mensagem)
        return {"Alerta enviado!!"}
    
    return {"Preço acima da média"}
```
